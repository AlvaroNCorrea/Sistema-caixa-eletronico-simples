```python
# Sitema de caixa eletronico
def depositar(saldo):
  print(f"Seu saldo é {saldo}")
  valor_deposito = float(input("Informe o valor que deseja depositar: "))
  saldo = saldo + valor_deposito
  return (f"seu novo saldo é: {saldo}")

def sacar(saldo):
  print(f"Seu saldo é {saldo}")
  valor_saque = float(input("Informe o valor que deseja sacar: "))
  if valor_saque > saldo:
    return("Impossivel realizar saque - saldo insuficiente")
  else:
    saldo = saldo - valor_saque
    return (f"seu novo saldo é: {saldo}")
```

```python
print("Seja bem vindo")
saldo = float(input("Por gentileza, informe seu saldo: "))
```

```python
acao = int(input("O que deseja fazer: 1 - ver saldo   2 - despositar   3 - sacar "))
if acao == 1:
  print(f"Seu saldo é: {saldo}")
elif acao == 2:
  print(depositar(saldo))
elif acao == 3:
  print(sacar(saldo))
else:
  print("Acao invalida")
```
